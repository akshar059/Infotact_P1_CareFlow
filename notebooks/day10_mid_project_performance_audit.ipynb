```python
import pandas as pd
import numpy as np

# Load processed datasets
cases_df = pd.read_csv('data/processed_new/careflow_cases_imputed.csv')
events_df = pd.read_csv('data/processed_new/careflow_events_scaled.csv')

print("=== DAY 10: MID-PROJECT PERFORMANCE AUDIT CHECKPOINT ===")
```

```python
# 1. Summary Statistics for Key Metrics
audit_summary = {
    'Total Cases': len(cases_df),
    'Total Events': len(events_df),
    'Avg Wait Time (min)': cases_df['Avg Wait Time Min'].mean() if 'Avg Wait Time Min' in cases_df else cases_df['assessment_to_discharge_min'].mean(),
    'Critical Cases Count': cases_df[cases_df['acuity'] <= 2].shape[0],
    'Unique Departments': cases_df['department'].nunique() if 'department' in cases_df else 'N/A'
}
```

```python
# 2. Display Audit Report
audit_df = pd.DataFrame(list(audit_summary.items()), columns=['Metric', 'Audit Value'])
print("\n--- PHASE 2 CHECKPOINT AUDIT SUMMARY ---")
print(audit_df.to_string(index=False))
```

```python
# 3. Save Audit Checkpoint
audit_df.to_csv('data/processed_new/day10_audit_summary.csv', index=False)
print("\n[SUCCESS] Day 10 Audit Summary saved successfully!")
```
