```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
# Ensure directory exists
os.makedirs('../data/processed_new', exist_ok=True)

# Load outlier-tagged dataset from Day 11
cases_df = pd.read_csv('../data/processed_new/careflow_cases_outliers.csv')

print('=== DAY 12: ADVANCED COHORT SEGMENTATION & BOTTLENECK DEEP-DIVE ===')
```

```python
# Determine the wait time column name dynamically
wait_col = (
    'Avg Wait Time Min'
    if 'Avg Wait Time Min' in cases_df
    else 'assessment_to_discharge_min'
)
```

```python
# 1. Segmenting Patients by Acuity & Wait Time Outliers
# Acuity <= 2 as High Acuity (Critical), > 2 as Low/Moderate Acuity
cases_df['cohort_segment'] = np.where(
    cases_df['acuity'] <= 2, 'High Acuity / Critical', 'Standard Acuity'
)
```

```python
# 2. Group-level Summary Statistics
segment_summary = (
    cases_df.groupby(['cohort_segment', 'is_wait_time_outlier'])[wait_col]
    .agg(['count', 'mean', 'median', 'max'])
    .reset_index()
)
print('\n--- COHORT SEGMENTATION & WAIT TIME ANALYSIS ---')
print(segment_summary.to_string(index=False))
```

```python
# 3. Save Segmented Dataset for Phase 3 Checkpoint
cases_df.to_csv(
    '../data/processed_new/careflow_cases_segmented.csv', index=False
)
print(
    '\n[SUCCESS] Saved segmented cohort dataset to'
    ' careflow_cases_segmented.csv'
)
```

### 📌 Day 12 Key Insights & Cohort Segmentation Summary

1. **Patient Cohort Categorization:** Segmented clinical cases into High-Acuity (Critical) and Standard Acuity groups to isolate high-risk operational paths.
2. **Bottleneck Analysis by Segment:** Evaluated mean and median wait times across outlier flags, highlighting where severe delay bottlenecks occur within specific clinical tiers.
3. **Pipeline Progression:** Exported the segmented dataset (`careflow_cases_segmented.csv`) as a key Phase 3 analytical artifact.
